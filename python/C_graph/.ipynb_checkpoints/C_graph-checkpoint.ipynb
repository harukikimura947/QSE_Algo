```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# パラメータのリスト（K, par_L, eta の順）
parameter_sets = [
    (2500, 0.0050, 1.5000),
    (3600, 0.0050, 1.8000),
    (4900, 0.0080, 1.5000),
    (6400, 0.0050, 1.5000),
    (8100, 0.0050, 1.8000),
    (10000, 0.0050, 1.5000),
    # 必要に応じて追加
]

# ベースのディレクトリ
base_path = "C://Users//kimura//github_repo//QSE_Algo//C//data//experiment"

# 各パラメータセットに対してグラフを描画
for K, par_L, eta in parameter_sets:
    # パスの構築
    dir_name = f"par_L={par_L:.4f},eta={eta:.4f}"
    file_path = os.path.join(base_path, f"K={K}", dir_name, "iteration_data.csv")

    # CSV読み込み
    try:
        df = pd.read_csv(file_path)
    except FileNotFoundError:
        print(f"ファイルが見つかりません: {file_path}")
        continue

    # グラフの描画
    plt.figure(figsize=(12, 8))

    plt.subplot(2, 2, 1)
    plt.semilogy(df['iteration'], df['max_diff'], label='max_diff')
    plt.title('max_diff')
    plt.xlabel('Iteration')
    plt.ylabel('max_relative_diff')
    plt.grid(True)
    plt.ylim(1e-10, None)  # 下限を固定

    plt.subplot(2, 2, 2)
    plt.semilogy(df['iteration'], df['R_max_diff'], label='R_max_diff', color='orange')
    plt.title('R_max_diff')
    plt.xlabel('Iteration')
    plt.ylabel('max_relative_diff')
    plt.grid(True)
    plt.ylim(1e-10, None)

    plt.subplot(2, 2, 3)
    plt.semilogy(df['iteration'], df['W_max_diff'], label='W_max_diff', color='green')
    plt.title('W_max_diff')
    plt.xlabel('Iteration')
    plt.ylabel('max_relative_diff')
    plt.grid(True)
    plt.ylim(1e-10, None)

    plt.subplot(2, 2, 4)
    plt.semilogy(df['iteration'], df['Z_diff'], label='Z_diff', color='red')
    plt.title('Z_diff')
    plt.xlabel('Iteration')
    plt.ylabel('relative_diff')
    plt.grid(True)
    plt.ylim(1e-10, None)

    # タイトルにパラメータを表示
    plt.suptitle(f"K={K}, par_L={par_L:.4f}, eta={eta:.4f}", fontsize=16)

    # レイアウト調整
    plt.tight_layout(rect=[0, 0, 1, 0.95])

    # グラフを表示（必要に応じて保存も可能）
    plt.show()
```
