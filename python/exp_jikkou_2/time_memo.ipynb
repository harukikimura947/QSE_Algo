```python
import matplotlib.pyplot as plt
import numpy as np

# フォント設定（Windows用の日本語対応フォント）
plt.rcParams['font.family'] = 'MS Gothic'

# データ
K_squared = [6250000, 12960000, 24010000, 40960000, 65610000, 100000000]
CPU_time = [17.888, 23.996, 95.460, 95.538, 213.975, 376.744]
iteration = [130, 79, 157, 93, 130, 150]
K_labels = [r'$K^2 = 50^4$', r'$K^2 = 60^4$', r'$K^2 = 70^4$', r'$K^2 = 80^4$', r'$K^2 = 90^4$', r'$K^2 = 100^4$']

# 1iterationあたりのCPU時間を計算
cpu_per_iter = np.array(CPU_time) / np.array(iteration)

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(K_squared, cpu_per_iter, marker='o', color='green', linestyle='-', linewidth=2)

# ラベル追加（点の右上）
for x, y in zip(K_squared, cpu_per_iter):
    plt.text(x + 1e6, y + 0.2, f'{y:.2f}', fontsize=9, color='black')

# 軸ラベル・タイトル
plt.xlabel(r'$K^2$（計算規模）', fontsize=14)
plt.ylabel('1 iteration あたりの CPU時間（秒）', fontsize=14)
plt.title(r'$K^2$ に対する 1 iteration あたりの CPU時間', fontsize=16)

plt.xticks(K_squared, K_labels, rotation=0)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.tight_layout()
plt.show()
```
